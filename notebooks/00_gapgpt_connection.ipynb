```python
from openai import OpenAI
from dotenv import load_dotenv
import os

load_dotenv()

test_client = OpenAI(
    api_key=os.getenv("GAPGPT_API_KEY"),
    base_url=os.getenv("GAPGPT_BASE_URL"),
    timeout=30.0,
    max_retries=0,
)

response = test_client.chat.completions.create(
    model=os.getenv("GAPGPT_MODEL"),
    messages=[
        {
            "role": "user",
            "content": "What is the weather in Berlin?",
        }
    ],
    tools=[
        {
            "type": "function",
            "function": {
                "name": "get_weather",
                "description": "Get the current weather for a city.",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "city": {
                            "type": "string",
                            "description": "The city name",
                        }
                    },
                    "required": ["city"],
                },
            },
        }
    ],
    tool_choice="required",
    max_tokens=100,
)

message = response.choices[0].message

print("Content:", message.content)
print("Tool calls:", message.tool_calls)
```

```python
from langchain_tavily import TavilySearch

search_tool = TavilySearch(
    max_results=5,
    topic="general",
    search_depth="basic",
)

results = search_tool.invoke({
    "query": "FastAPI vs Django performance comparison"
})

print(results)
```

```python
print("Number of results:", len(results["results"]))

for index, item in enumerate(results["results"], start=1):
    print(f"\nResult {index}")
    print("Title:", item["title"])
    print("URL:", item["url"])
    print("Score:", item["score"])
```
