```python
import os
from dotenv import load_dotenv
from openai import OpenAI

# Load environment variables
load_dotenv()

# Read GapGPT configuration
api_key = os.getenv("GAPGPT_API_KEY")
base_url = os.getenv("GAPGPT_BASE_URL")
model = os.getenv("GAPGPT_MODEL")

# Check configuration
if not api_key:
    raise ValueError("GAPGPT_API_KEY was not found in .env")

if not base_url or not model:
    raise ValueError("GapGPT configuration is incomplete")

print("Configuration loaded successfully!")
print("Model:", model)
```

```python

client = OpenAI(
    api_key=api_key,
    base_url=base_url
)

response = client.responses.create(
    model=model,
    input="Explain what an AI research agent is in three sentences."
)

print(response.output_text)
```
